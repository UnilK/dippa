```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from medium_attenuation import *
```

```python

f = np.linspace(100, 8000, 2 ** 10)
dist = np.linspace(1, 200, 200)
T = 12
hr = 0.5
alpha_iso = air_attenuation(T, f, hr) 

dBToAmp = lambda dB: np.pow(10, dB / 20)
ampTodB = lambda amp: 20 * np.log10(amp)

alpha = alpha_iso.freq
vf, vd = np.meshgrid(f, dist)

att = alpha * vd

plt.rc('font', **{'size': 16})
plt.figure(figsize=(10, 6))
pcm = plt.pcolormesh(vf, vd, att, cmap='GnBu', shading='auto')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Distance (m)')

cs = plt.contour(vf, vd, att, colors='k', linewidths=0.6, levels=10)
plt.clabel(cs, inline=True, fontsize=8)

cbar = plt.colorbar(pcm)
cbar.set_label('Attenuation (dB)')
# plt.title("Air Attenuation @ 10°C, 50% relative humidity")
plt.tight_layout()
plt.show()
```
